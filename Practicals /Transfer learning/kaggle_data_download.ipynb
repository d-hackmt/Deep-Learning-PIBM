```python
# Set destination directory in Google Drive
drive_path = '/content/drive/MyDrive/Kaggle/dogs-vs-cats'
os.makedirs(drive_path, exist_ok=True)

# Download the competition zip file to Google Drive
!kaggle competitions download -c dogs-vs-cats -p "{drive_path}"
```

```python
# Extract dogs-vs-cats.zip
from zipfile import ZipFile

dataset_zip = os.path.join(drive_path, 'dogs-vs-cats.zip')

with ZipFile(dataset_zip, 'r') as zip_ref:
    zip_ref.extractall(drive_path)
    print('Main dataset extracted to Google Drive.')

# Extract train.zip inside extracted folder
train_zip = os.path.join(drive_path, 'train.zip')
with ZipFile(train_zip, 'r') as zip_ref:
    zip_ref.extractall(os.path.join(drive_path, 'train'))
    print('Train images extracted to Google Drive.')


```
